```python
n = int(input("Enter number: "))

reverse = 0

for digit in str(n)[::-1]:
    digit = int(digit)
    reverse = reverse * 10 + digit

print("Reverse =", reverse)
```

```python
n = int(input("Enter number: "))

reverse = 0
while n > 0:
    digit = n % 10 
    reverse = reverse * 10 + digit
    n = n//10 
print(reverse)
```

```python
n = 12133
new_n = n
reverse = 0 
while n > 0:
    digit = n % 10 
    reverse = reverse * 10 + digit 
    n = n // 10 
if new_n == reverse:
    print("Palindrom")
else:
    print("not palindrom")
```

```python
n = int(input("Enter number: "))
count = 0
while n > 0:
    digit = n % 10
    if digit == 0:
        count += 1
    n = n // 10

print("Count of zeros =", count)
```
