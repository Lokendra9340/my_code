```python
def even(a):
    if a % 2 == 0:
        print("Even")
x  = even(6)
print(x)
```

```python
def addJ(a):
    if a > 0:
        print("Positive")
    elif a < 0:
        print("negetive")
    else:
        ("Zero")
addJ(-5)
```
